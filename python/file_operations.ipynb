```python
with open(r'example_files/data.txt', 'r') as file:
    data = file.read()
    print(data)
```

```python
with open(r'example_files/data.txt', 'r') as file:
    for line in file:
        print(line.strip())
```

```python
with open(r'example_files/data.txt', 'w') as file: # overwriting
    file.write("This is a new line of text.\n")
    file.write("This is another line of text.\n")
```

```python
with open(r'example_files/data.txt', 'a') as file: # appending
    file.write("This is a new line of text 1.\n")
    file.write("This is another line of text 1.\n")
```

```python
list_lines = ["This is line 1.\n", "This is line 2.\n", "This is line 3.\n"]
with open(r'example_files/data.txt', 'a') as file:
    file.writelines(list_lines)
    
```
