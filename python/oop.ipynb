```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance
    def deposit(self,amount):
        self.balance += amount
        print(f"Added {amount} to the balance. The new balance is {self.balance}")
    def withdraw(self,amount):
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount
        print(f"Withdrew {amount} from the balance. The new balance is {self.balance}")
```

```python
account = BankAccount("John",1000)
```

```python
account.deposit(50000000000)
```

```python
account.withdraw(50000000000)
```

### Inherticance

```python
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"The person is driving the car with {self.windows} windows, {self.doors} doors, and a {self.enginetype} engine.")
```

```python
car1 = Car(4,4,"v8-petrol")
car1.drive()
```

```python
class Tesla(Car):
    def __init__(self,windows,doors,enginetype,is_self_driving):
        super().__init__(windows,doors,enginetype)
        self.is_self_driving = is_self_driving

    def self_drive(self):
        if self.is_self_driving:
            print("The Tesla is driving itself.")
        else:
            print("The Tesla is not self-driving.")
```

```python
Tesla1 = Tesla(4,4,"electric",True)
Tesla1.drive()
Tesla1.self_drive()
```

```python
### Multiple Inheritance
class Animal:
    def __init__(self,name):
        self.name = name
    def speak(self):
        print("Sub Classes must implement this method.")
        print(f"{self.name} makes a sound.")

class Pet:
    def __init__(self,owner):
        self.owner = owner
    def play(self):
        print(f"{self.owner}'s pet is playing.")   

class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    def speak(self):
        print(f"{self.name} says Woof!")
```

```python
dog1 = Dog("Buddy","Alice")
dog1.speak()
dog1.play()
```

### Polymorphism

```python
#Method Overriding
class Cat(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    def speak(self):
        print(f"{self.name} says Meow!")

    def play(self):
        print(f"{self.owner}'s cat is playing with a ball of yarn.")
```

```python
cat1 = Cat("Whiskers","Bob")
cat1.speak()
```

```python
def animal_sound(animal):
    animal.speak()
```

```python
print("Dog sound:")
animal_sound(dog1)
print("Cat sound:")
animal_sound(cat1)  
```

```python
class Shape:
    def __init__(self):
        pass
    def area(self):
        print("Area calculation is not implemented for the base class.")

class Rectangle(Shape):
    def __init__(self,width,height):
        super().__init__()
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height 
class circle(Shape):
    def __init__(self,radius):
        super().__init__()
        self.radius = radius
    def area(self):
        import math
        return math.pi * self.radius ** 2
```

```python
def print_area(shape):
    print(f"The area of the shape is: {shape.area()}")
```

```python
print_area(Rectangle(5,10))
print_area(circle(7))   
```

```python
# Polymorphism with abstract base classes
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
class Cars(Vehicle):
    def start_engine(self):
        print("Starting the car's engine.") 

class Motorcycles(Vehicle):
    def start_engine(self):
        print("Starting the motorcycle's engine.")
```

```python
car = Cars()
car.start_engine()
mc = Motorcycles()
mc.start_engine()
```

### Encapsulation

```python
class Person:
    def __init__(self,name,age,gender,salary):
        self.name = name # Public attribute
        self.age = age
        self._gender = gender  # Protected attribute
        self.__salary = salary  # Private attribute  
    def greet(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")
    def get_salary(self):
        return self.__salary
```

```python
person1 = Person("Alice",30,"F",50000)
person1.age = 31  # Modifying the public attribute
person1.greet()  # Output: Hello, my name is Alice and I am 31 years old.
person1._gender = "Female"  # Modifying the protected attribute
print(person1.__salary)  # Output: 60000 (but this is not the actual private attribute)
person1.get_salary()  # Output: 50000 (the actual private attribute)
```

```python
class Person:
    def __init__(self,name,age,gender,salary):
        self.name = name # Public attribute
        self.age = age
        self._gender = gender  # Protected attribute
        self.__salary = salary  # Private attribute  

    # getters and setters methods
    def get_name(self):
        return self.name
    def set_name(self,name):
        self.name = name
    def get_age(self):
        return self.age
    def set_age(self,age):
        self.age = age
    def get_gender(self):
        return self._gender
    def set_gender(self,gender):
        self._gender = gender
    def get_salary(self):
        return self.__salary    
    def set_salary(self,salary):
        if salary > 0 or salary != self.__salary:
            self.__salary = salary  
```

```python
person1 = Person("Alice",30,"F",50000)
person1.set_name("Bob")
person1.set_age(35)
print(person1.get_name())  # Output: Bob
print(person1.get_age())   # Output: 35     
print(person1.get_gender())  # Output: F
print(person1.get_salary())  # Output: 50000
person1.set_salary(60000)
print(person1.get_salary())  # Output: 60000
person1.set_gender("M")
print(person1.get_gender())  # Output: M
```

### Abstraction

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is driving.")

    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Starting the car's engine.")
```

```python
def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car = Car()
operate_vehicle(car)  # Output: Starting the car's engine. The vehicle is driving.
```

```python
# Magic Methods or Dunder Methods

class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person(name={self.name}, age={self.age})"

    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

    def __len__(self):
        return self.age

    def __add__(self,other):
        if isinstance(other,Person):
            return Person(self.name + " & " + other.name,self.age + other.age)
        return NotImplemented

```

```python
person1 = Person("Alice",30)
print(person1)
```
