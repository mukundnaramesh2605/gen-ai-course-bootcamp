```python
corpus = """Hello Welcome, to the tutorial on NLP.
Please make sure to install the required libraries before running the code!
To become a master in NLP, you need to practice a lot and read a lot of research papers!
I'm AI enthusiast and I love to explore the world of AI and NLP. I hope you enjoy this tutorial and learn a lot from it!"""
```

```python
print(corpus)
```

```python
## Tokenization is the process of breaking down a text into smaller units called tokens. These tokens can be words, phrases, or even sentences. Tokenization is an essential step in natural language processing (NLP) as it helps in understanding the structure and meaning of the text.
import nltk
from nltk.tokenize import sent_tokenize
nltk.download('punkt_tab')
documents = sent_tokenize(corpus)
for i, sentence in enumerate(documents):
    print(f"{i + 1}: {sentence}")
```

```python
type(documents)
```

```python
# Converting paras into words
from nltk.tokenize import word_tokenize
words = word_tokenize(corpus)
for word in words:
    print(word)
```

```python
words_documents = [word_tokenize(sentence) for sentence in documents]
for i, sentence in enumerate(words_documents):
    print(f"Sentence {i + 1}: {sentence}")
```

```python
from nltk.tokenize import wordpunct_tokenize
words = wordpunct_tokenize(corpus)
for word in words:
    print(word)
```

```python
from nltk.tokenize.treebank import TreebankWordTokenizer
words = TreebankWordTokenizer().tokenize(corpus)
for word in words:
    print(word)
```
